```python
###problem1
TP=50
FP=8
TN=32
FN=10
#1--total prediction
total=TP+FP+TN+FN
print("Total Predictions: ",total)
#2--calculate accuracy, precision,recall and f1 score
accuracy=(TP+TN)/total
print("Accuracy: ",accuracy)

precision=TP/(TP+FP)
print("Precision: ",precision)

recall=TP/(TP+FN)
print("Recall: ",recall)

f1=2*precision*recall/(precision+recall)
print("F1 Score: ",f1)
```

```python
###problem2

#given values
SSR = 420
SSE = 180

SST = SSR + SSE          #total squared error
R_squared = SSR / SST    #R^2

#resuls-SST and R^2
print("SST (Total sum of squared error):", SST)
print("R^2 (Coefficient of Determination):", R_squared)

#check if the model explain more than 70% variance
if R_squared > 0.70:
    print("The model explains more than 70% of the variance.")
else:
    print("The model explains less than or equal to 70% of the variance.")
```

```python
###problem3

#given squared errors stored in a list
squared_errors = [2, 4, 1, 9, 4]
#no. of observations
n = len(squared_errors)

#calculate MSE
mse = sum(squared_errors) / n
#results
print("Squared Errors:", squared_errors)
print("Mean Squared Error (MSE):", mse)
```

```python
###problem4

#given values
e2 = 9         #squared residual
MSE = 3        #mean squared residual
p = 3          #no. of prediction
hii = 0.2      #leverage
n = 25         #sample size

#cook's D distance
Di = (e2 / (p * MSE)) * (hii / ((1 - hii) ** 2))
#threshold
threshold = 4 / n
#results
print("Cook's Distance (Di):", Di)
print("Threshold (4/n):", threshold)

#check if observation is influential
if Di > threshold:
    print("The observation is influential.")
else:
    print("The observation is NOT influential.")
```
