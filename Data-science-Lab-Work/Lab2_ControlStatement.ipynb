```python
### indentation
### control statement
##  to check if a number is positive

num = int(input("Enter your number : "))
if num>0:
  print("The number is positive ")

elif num<0:
  print("The number is negetive ")

elif num==0:
  print("The number is zero ")
```

```python
### whether given number is odd or even
num = int(input("Enter your number : "))
if (num%2)==0:
  print("The number is even ")

else :
  print("The number is odd ")
```

```python
### python code to print multiplication table
for i in range(1,11) :
  for j in range(1,11) :
    print("{:5}".format(i*j),end=' ')
  print()
```

```python
### while loop

count=0
while (count<5):
  print("My count is ",count)
  count=count+1
```

```python
### enumerate
students = ("Clifford","Angelina","Kishore","Aarav","Rehan","Ganesh","Jena")
for ind,stu in enumerate(students):
  print(ind+1,stu)
```

```python
### loops in dictionary
students = {"Abhijeet":90,"Gurneet":91,"Marc":89,"Jena":88}
for i in students:
  marks=students[i]
  #print(marks)
  print("Student %s has scored %d"%(i,marks))
```

```python
print(students.values())
(students.keys())
for i in students.values():
  print(i)
for k,v in students.items():
  print(k,v)
```

```python
nums = [0,1,2,3,4,5]
sq_n = []
for i in nums:
  sq_n.append(i**2)
  print(sq_n)
```

```python
### comprehension in python
print(x**2 for x in nums)
```

```python
cal = [89,90,78,77,88,67,65,89,89,90]
for marks in cal:
  if marks>70:
    print(marks)
```

```python
h=[marks for marks in cal if marks>70]
print(h)
```

```python
print([marks for marks in cal if marks%2==0])
```

```python
main_list=[[1,2,3],[4,5,6],[7,8,9]]
consise_list=[]
for sublist in main_list:
  #print(sublist)
  for el in sublist:
    consise_list.append(el)
print(consise_list)
```

```python
print([ele for slist in main_list for ele in slist])  ### comprehension
```

```python
mlist=[["Mercury", "Venus", "Earth"],
       ["Mars", "Jupiter", "Saturn"],
       ["Uranus", "Neptune", "Pluto"]]
consise_list=[]
for slist in mlist:
  for el in slist:
    if len(el)<6:
      consise_list.append(el)
print(consise_list)

#Expected Output:
    #["Venus", "Earth", "Mars", "Pluto"]
```
